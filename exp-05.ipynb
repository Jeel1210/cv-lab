# Experiment 5 - Edge Detection with the Sobel Operator

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from numpy.lib.stride_tricks import sliding_window_view
from skimage import data
```

```python
img = data.camera()
smooth = cv2.GaussianBlur(img, (5, 5), 1.2)

KX = np.array([[-1, 0, 1],
               [-2, 0, 2],
               [-1, 0, 1]], dtype=np.float64)
KY = KX.T

def conv3(src, kern):
    padded = np.pad(src.astype(np.float64), 1, mode="edge")
    win = sliding_window_view(padded, (3, 3))
    return np.einsum("ijkl,kl->ij", win, kern)

gx = conv3(smooth, KX)
gy = conv3(smooth, KY)
mag = np.hypot(gx, gy)
mag = (mag / mag.max() * 255).astype(np.uint8)

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, im, t in zip(axes, [img, np.abs(gx), np.abs(gy), mag], ["input", "|Gx|", "|Gy|", "magnitude"]):
    ax.imshow(im, cmap="gray")
    ax.set_title(t)
    ax.axis("off")
plt.show()
```

```python
cx = cv2.Sobel(smooth, cv2.CV_64F, 1, 0, ksize=3, borderType=cv2.BORDER_REPLICATE)
cy = cv2.Sobel(smooth, cv2.CV_64F, 0, 1, ksize=3, borderType=cv2.BORDER_REPLICATE)
print("max diff Gx:", np.abs(cx - gx).max())
print("max diff Gy:", np.abs(cy - gy).max())
```
