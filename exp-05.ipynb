# Experiment 5 - Edge Detection with the Sobel Operator

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from numpy.lib.stride_tricks import sliding_window_view
from skimage import data
```

```python
img = data.camera()
smooth = cv2.GaussianBlur(img, (5, 5), 1.2)

KX = np.array([[-1, 0, 1],
               [-2, 0, 2],
               [-1, 0, 1]], dtype=np.float64)
KY = KX.T

def conv3(src, kern):
    padded = np.pad(src.astype(np.float64), 1, mode="edge")
    win = sliding_window_view(padded, (3, 3))
    return np.einsum("ijkl,kl->ij", win, kern)

gx = conv3(smooth, KX)
gy = conv3(smooth, KY)
mag = np.hypot(gx, gy)
mag = (mag / mag.max() * 255).astype(np.uint8)

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, im, t in zip(axes, [img, np.abs(gx), np.abs(gy), mag], ["input", "|Gx|", "|Gy|", "magnitude"]):
    ax.imshow(im, cmap="gray")
    ax.set_title(t)
    ax.axis("off")
plt.show()
```

```python
cx = cv2.Sobel(smooth, cv2.CV_64F, 1, 0, ksize=3, borderType=cv2.BORDER_REPLICATE)
cy = cv2.Sobel(smooth, cv2.CV_64F, 0, 1, ksize=3, borderType=cv2.BORDER_REPLICATE)
print("max diff Gx:", np.abs(cx - gx).max())
print("max diff Gy:", np.abs(cy - gy).max())
```

```python
theta = np.degrees(np.arctan2(gy, gx))

fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for ax, T in zip(axes, [30, 60, 100]):
    ax.imshow(mag > T, cmap="gray")
    ax.set_title(f"edges, threshold = {T}")
    ax.axis("off")
plt.show()

plt.figure(figsize=(6, 5))
plt.imshow(np.where(mag > 60, theta, np.nan), cmap="hsv")
plt.colorbar(label="gradient direction (deg)")
plt.title("direction on strong edges")
plt.axis("off")
plt.show()
```

```python
ksizes = [3, 5, 7]
fig, axes = plt.subplots(1, len(ksizes), figsize=(16, 4))
for ax, k in zip(axes, ksizes):
    sx = cv2.Sobel(smooth, cv2.CV_64F, 1, 0, ksize=k)
    sy = cv2.Sobel(smooth, cv2.CV_64F, 0, 1, ksize=k)
    m = cv2.magnitude(sx, sy)
    ax.imshow(m / m.max(), cmap="gray")
    ax.set_title(f"cv2.Sobel ksize={k}")
    ax.axis("off")
plt.show()
```
