# Experiment 2 - Salt and Pepper Noise

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
img = cv2.cvtColor(data.chelsea(), cv2.COLOR_RGB2GRAY)
print(img.shape, img.dtype)

plt.imshow(img, cmap="gray")
plt.axis("off")
plt.show()
```

```python
rng = np.random.default_rng(7)

def salt_pepper(src, amount=0.1, salt_ratio=0.5):
    out = src.copy()
    n = int(amount * src.size)
    n_salt = int(n * salt_ratio)
    idx = rng.choice(src.size, size=n, replace=False)
    flat = out.reshape(-1)
    flat[idx[:n_salt]] = 255
    flat[idx[n_salt:]] = 0
    return out
```

```python
levels = [0.02, 0.1, 0.25, 0.4]

fig, axes = plt.subplots(1, len(levels) + 1, figsize=(18, 4))
axes[0].imshow(img, cmap="gray")
axes[0].set_title("clean")
for ax, p in zip(axes[1:], levels):
    noisy = salt_pepper(img, p)
    ax.imshow(noisy, cmap="gray")
    ax.set_title(f"amount = {p}")
for ax in axes:
    ax.axis("off")
plt.show()
```

```python
noisy = salt_pepper(img, 0.15)
print("black px:", int((noisy == 0).sum()))
print("white px:", int((noisy == 255).sum()))
print("changed fraction:", round(float((noisy != img).mean()), 4))
```
