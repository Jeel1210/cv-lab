# Experiment 1 - Image Basics with OpenCV

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

print(f"opencv {cv2.__version__} | numpy {np.__version__}")
```

```python
IMG_PATH = "cat.jpg"
cv2.imwrite(IMG_PATH, cv2.cvtColor(data.chelsea(), cv2.COLOR_RGB2BGR))

bgr = cv2.imread(IMG_PATH)
assert bgr is not None, f"could not read {IMG_PATH}"

print("shape:", bgr.shape)
print("dtype:", bgr.dtype)
print("size :", bgr.size)
```

```python
def show(img, title="", cmap=None, ax=None):
    if ax is None:
        ax = plt.gca()
    if img.ndim == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    ax.imshow(img, cmap=cmap)
    ax.set_title(title)
    ax.axis("off")

plt.figure(figsize=(7, 5))
show(bgr, "original")
plt.show()
```

```python
rows, cols, ch = bgr.shape
print(f"rows={rows} cols={cols} channels={ch}")

y, x = 150, 220
b, g, r = bgr[y, x]
print(f"pixel at ({y}, {x}) -> B={b} G={g} R={r}")
```

```python
edited = bgr.copy()
edited[y - 5:y + 5, x - 5:x + 5] = (255, 0, 0)

crop = bgr[60:260, 120:380]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
show(edited, "pixel block changed", ax=axes[0])
show(crop, f"crop {crop.shape[:2]}", ax=axes[1])
plt.show()
```

```python
gray = cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)
small = cv2.resize(bgr, (320, 240), interpolation=cv2.INTER_AREA)
print("gray:", gray.shape, "| resized:", small.shape)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
show(gray, "grayscale", cmap="gray", ax=axes[0])
show(small, "resized 320x240", ax=axes[1])
plt.show()
```

```python
variants = {
    "flip horizontal": cv2.flip(bgr, 1),
    "flip vertical": cv2.flip(bgr, 0),
    "flip both": cv2.flip(bgr, -1),
    "rotate 90 ccw": cv2.rotate(bgr, cv2.ROTATE_90_COUNTERCLOCKWISE),
}

fig, axes = plt.subplots(1, len(variants), figsize=(16, 4))
for ax, (name, im) in zip(axes, variants.items()):
    show(im, name, ax=ax)
plt.show()
```

```python
canvas = bgr.copy()
cv2.rectangle(canvas, (130, 40), (330, 230), (0, 200, 255), 2)
cv2.circle(canvas, (230, 135), 40, (0, 255, 0), 2)
cv2.line(canvas, (0, rows - 1), (cols - 1, 0), (255, 0, 255), 1)
cv2.putText(canvas, "CV Lab", (20, 40), cv2.FONT_HERSHEY_DUPLEX, 1.1, (255, 255, 255), 2, cv2.LINE_AA)

plt.figure(figsize=(7, 5))
show(canvas, "shapes and text")
plt.show()
```
